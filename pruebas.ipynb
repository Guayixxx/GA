```python
from PIL import Image
```

```python
def convertir_a_grises(ruta_entrada, ruta_salida):
    """
    Convierte una imagen a escala de grises.
    
    :param ruta_entrada: Ruta de la imagen de entrada (color).
    :param ruta_salida: Ruta donde se guardará la imagen en escala de grises.
    """
    try:
        # Cargar la imagen
        imagen = Image.open(ruta_entrada)
        # Convertir a escala de grises
        imagen_grises = imagen.convert('L')
        # Guardar la imagen
        imagen_grises.save(ruta_salida)
        print(f"Imagen convertida a escala de grises y guardada en: {ruta_salida}")
    except Exception as e:
        print(f"Error al convertir a grises: {e}")
```

```python
def convertir_a_color(ruta_entrada, ruta_salida):
    """
    Convierte una imagen en escala de grises a color (replicando tonos de gris en RGB).
    
    :param ruta_entrada: Ruta de la imagen de entrada (escala de grises).
    :param ruta_salida: Ruta donde se guardará la imagen en color.
    """
    try:
        # Cargar la imagen
        imagen = Image.open(ruta_entrada)
        # Convertir a color (RGB)
        imagen_color = imagen.convert('RGB')
        # Guardar la imagen
        imagen_color.save(ruta_salida)
        print(f"Imagen convertida a color y guardada en: {ruta_salida}")
    except Exception as e:
        print(f"Error al convertir a color: {e}")
```

```python
# Ejemplo de uso
ruta_original = 'Foto.jpg'  # Cambia por la ruta de tu imagen
ruta_grises = 'imagen_grises.jpg'
ruta_color = 'imagen_color.jpg'
```

```python
# Convertir a grises
convertir_a_grises(ruta_original, ruta_grises)
```

```python
# Convertir de grises a color
convertir_a_color(ruta_grises, ruta_color)
```

```python
from PIL import Image

def obtener_valores_unicos_rgb(ruta_imagen):
    """
    Obtiene los valores únicos RGB de los píxeles de una imagen.
    
    :param ruta_imagen: Ruta de la imagen de entrada.
    :return: Un conjunto de tuplas con los valores RGB únicos.
    """
    try:
        # Cargar la imagen
        imagen = Image.open(ruta_imagen)
        
        # Convertir la imagen a RGB (en caso de que no lo sea)
        imagen_rgb = imagen.convert('RGB')
        
        # Obtener los píxeles de la imagen
        pixeles = imagen_rgb.getdata()
        
        # Usar un conjunto para obtener los valores únicos de los píxeles (sin duplicados)
        valores_unicos_rgb = set(pixeles)
        
        # Mostrar la cantidad de valores únicos encontrados
        print(f"Se encontraron {len(valores_unicos_rgb)} colores únicos en la imagen.")
        
        return valores_unicos_rgb
    
    except Exception as e:
        print(f"Error al obtener los valores únicos RGB: {e}")
        return set()
```

```python
# Ejemplo de uso
ruta_imagen = 'Foto.jpg'  # Cambia por la ruta de tu imagen

valores_unicos = obtener_valores_unicos_rgb(ruta_imagen)
print(valores_unicos)  # Mostrar los valores únicos RGB
```

```python
import numpy as np
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

def normalizar_colores(ruta_imagen, n_colores=4):
    """
    Normaliza los colores de una imagen a un número reducido de colores usando k-means clustering.
    
    :param ruta_imagen: Ruta de la imagen de entrada.
    :param n_colores: Número de colores que se desean obtener (por defecto 4).
    :return: Una nueva imagen con los colores normalizados.
    """
    try:
        # Cargar la imagen
        imagen = Image.open(ruta_imagen)
        
        # Convertir la imagen a RGB (en caso de que no lo sea)
        imagen_rgb = imagen.convert('RGB')
        
        # Obtener los datos de los píxeles
        pixeles = np.array(imagen_rgb).reshape(-1, 3)
        
        # Usar KMeans para agrupar los colores
        kmeans = KMeans(n_clusters=n_colores, random_state=42)
        kmeans.fit(pixeles)
        
        # Obtener los centroides (colores promedio) de los clusters
        colores_reducidos = kmeans.cluster_centers_.astype(int)
        
        # Asignar a cada píxel el color del centroide más cercano
        etiquetas = kmeans.predict(pixeles)
        pixeles_normalizados = colores_reducidos[etiquetas]
        
        # Crear una nueva imagen con los colores normalizados
        nueva_imagen = pixeles_normalizados.reshape(imagen_rgb.size[1], imagen_rgb.size[0], 3)
        imagen_normalizada = Image.fromarray(nueva_imagen.astype('uint8'))
        
        # Mostrar los colores representativos
        print(f"Colores normalizados: {colores_reducidos}")
        
        # Mostrar la imagen resultante
        imagen_normalizada.show()
        
        # Guardar la nueva imagen
        ruta_salida = 'imagen_normalizada.jpg'
        imagen_normalizada.save(ruta_salida)
        print(f"Imagen normalizada guardada en: {ruta_salida}")
        
        return imagen_normalizada
    
    except Exception as e:
        print(f"Error al normalizar los colores: {e}")
        return None
```

```python
# Ejemplo de uso
ruta_imagen = 'Foto.jpg'  # Cambia por la ruta de tu imagen
imagen_normalizada = normalizar_colores(ruta_imagen, n_colores=4)
```
